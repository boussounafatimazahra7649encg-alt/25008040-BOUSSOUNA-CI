```python

# ============================================================
# PARTIE 3 : EXPLORATION, CONTROLE ET VISUALISATION DES DONNEES
# Projet : Prévision des exportations agricoles du Maroc
# Source : UN Comtrade
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import unicodedata
import re

# 1. IDENTIFICATION DES COLONNES
# Cette fonction permet de reconnaître les colonnes
# même si leurs noms sont en français ou en anglais.

def normaliser(nom):
    nom = str(nom).lower().strip()
    nom = unicodedata.normalize('NFKD', nom)
    nom = ''.join(c for c in nom if not unicodedata.combining(c))
    return re.sub(r'[^a-z0-9]', '', nom)

def trouver_colonne(candidats):
    colonnes = {normaliser(c): c for c in df.columns}
    for candidat in candidats:
        if normaliser(candidat) in colonnes:
            return colonnes[normaliser(candidat)]
    raise ValueError(
        f"Colonne introuvable parmi {candidats}. "
        f"Colonnes disponibles : {df.columns.tolist()}"
    )

col_annee = trouver_colonne(
    ['Period', 'Période', 'Annee', 'Year']
)

col_produit = trouver_colonne(
    ['Commodity', 'Produit', 'Commodity Code', 'Code SH', 'cmdCode']
)

col_valeur = trouver_colonne(
    ['Trade Value (US$)', 'Valeur commerciale (US $)',
     'TradeValue', 'Valeur commerciale', 'primaryValue']
)

# Le poids net est facultatif
candidats_poids = [
    normaliser(c) for c in
    ['Netweight (kg)', 'Poids net (kg)', 'Netweight']
]
col_poids = next(
    (c for c in df.columns if normaliser(c) in candidats_poids),
    None
)

# Copie de travail
data = df.copy()

# 2. PREPARATION DES VARIABLES
data[col_annee] = pd.to_numeric(data[col_annee], errors='coerce')

def convertir_nombre(serie):
    if pd.api.types.is_numeric_dtype(serie):
        return pd.to_numeric(serie, errors='coerce')
    serie = serie.astype(str).str.replace(
        r'[\s\u00a0]', '', regex=True
    ).str.replace(',', '.', regex=False)
    return pd.to_numeric(serie, errors='coerce')

data[col_valeur] = convertir_nombre(data[col_valeur])

if col_poids is not None:
    data[col_poids] = convertir_nombre(data[col_poids])

# Conserver les codes SH comme texte pour préserver les zéros
data[col_produit] = data[col_produit].astype(str).str.strip()

print("===== DIMENSIONS DU DATASET =====")
print("Nombre de lignes :", data.shape[0])
print("Nombre de colonnes :", data.shape[1])

print("\n===== APERCU DES DONNEES =====")
display(data.head(10))

print("\n===== TYPES DES VARIABLES =====")
print(data.dtypes)

print("\n===== STATISTIQUES DESCRIPTIVES =====")
display(data.describe(include='all').T)


# 3. VERIFICATION DES VALEURS MANQUANTES
print("\n===== VALEURS MANQUANTES =====")

manquants = pd.DataFrame({
    'Nombre_manquants': data.isna().sum(),
    'Pourcentage (%)': data.isna().mean() * 100
})

display(manquants.sort_values(
    'Pourcentage (%)', ascending=False
).round(2))


# 4. VERIFICATION DES DOUBLONS
# Clé demandée : pays - partenaire - produit - période

colonnes_cle = []

for candidats in [
    ['Reporter', 'Déclarant'],
    ['Partner', 'Partenaire'],
    ['Commodity', 'Produit', 'Commodity Code', 'Code SH', 'cmdCode'],
    ['Period', 'Période', 'Annee', 'Year']
]:
    for candidat in candidats:
        trouve = next(
            (c for c in data.columns
             if normaliser(c) == normaliser(candidat)),
            None
        )
        if trouve is not None:
            colonnes_cle.append(trouve)
            break

print("\n===== CONTROLE DES DOUBLONS =====")

doublons = data.duplicated(subset=colonnes_cle, keep=False)

print("Nombre de lignes concernées par les doublons :",
      doublons.sum())

if doublons.any():
    print("Exemples de doublons :")
    display(data.loc[doublons, colonnes_cle].head(10))
else:
    print("Aucun doublon détecté selon la clé retenue.")


# 5. CONTROLE DES UNITES ET DES VALEURS
print("\n===== CONTROLE DES UNITES =====")

for nom in ['Trade Flow', 'Flux commercial',
            'Quantity Unit', 'Unité',
            'Netweight Unit', 'Unité de poids']:
    for c in data.columns:
        if normaliser(c) == normaliser(nom):
            print(f"\n{c} :")
            print(data[c].value_counts(dropna=False))

print("\nValeurs commerciales négatives :",
      (data[col_valeur] < 0).sum())

if col_poids is not None:
    print("Poids nets négatifs :", (data[col_poids] < 0).sum())


# 6. PREPARATION POUR LES GRAPHIQUES
data_graph = data.dropna(
    subset=[col_annee, col_produit, col_valeur]
).copy()

data_graph[col_annee] = data_graph[col_annee].astype(int)

# 7. FIGURE 1 : EVOLUTION GLOBALE DES EXPORTATIONS

exportations_annuelles = (
    data_graph.groupby(col_annee)[col_valeur]
    .sum()
    .sort_index()
)

plt.figure(figsize=(11, 5))
plt.plot(
    exportations_annuelles.index,
    exportations_annuelles.values / 1e6,
    marker='o',
    linewidth=2.5,
    color='royalblue',
    label='Exportations'
)

plt.title("Figure 1 : Evolution annuelle des exportations agricoles marocaines",
          fontsize=13)
plt.xlabel("Année")
plt.ylabel("Valeur des exportations (millions USD)")
plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()


# 8. FIGURE 2 : COMPARAISON DES PRODUITS

exportations_produits = (
    data_graph.groupby([col_annee, col_produit])[col_valeur]
    .sum()
    .unstack()
    .sort_index()
)

plt.figure(figsize=(12, 6))

for produit in exportations_produits.columns:
    plt.plot(
        exportations_produits.index,
        exportations_produits[produit] / 1e6,
        marker='o',
        linewidth=2,
        label=str(produit)
    )

plt.title("Figure 2 : Evolution des exportations par produit agricole",
          fontsize=13)
plt.xlabel("Année")
plt.ylabel("Valeur des exportations (millions USD)")
plt.grid(alpha=0.3)
plt.legend(title="Code SH")
plt.tight_layout()
plt.show()


# 9. FIGURE 3 : COMPARAISON DU POIDS NET PAR PRODUIT

if col_poids is not None:

    poids_produits = (
        data_graph.groupby(col_produit)[col_poids]
        .sum()
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(11, 5))
    poids_produits.plot(
        kind='bar',
        color=plt.cm.Set2(
            np.linspace(0, 1, len(poids_produits))
        )
    )

    plt.title("Figure 3 : Poids net total exporté par produit",
              fontsize=13)
    plt.xlabel("Code SH du produit")
    plt.ylabel("Poids net (kg)")
    plt.xticks(rotation=45)
    plt.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.show()

else:
    print("Figure 3 non réalisée : colonne du poids net absente.")

print("\n===== FIN DE L'EXPLORATION =====")

```

```python

# ============================================================
# PARTIE 4 : PREPARATION ET DECOUPAGE SANS FUITE DE DONNEES
# ============================================================

import pandas as pd
import numpy as np

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# ------------------------------------------------------------
# 1. COPIE DE LA BASE
# ------------------------------------------------------------

data = df.copy()

print("Dimensions initiales :", data.shape)

# ------------------------------------------------------------
# 2. FILTRAGE DES DONNEES (ADAPTE AU DATASET REEL)
# ------------------------------------------------------------

# On filtre par la colonne 'flowCode' s'il y a plusieurs flux,
# mais on s'assure d'abord de regarder ce qui est présent dans le dataset.
# Si 'flowCode' ou 'flowDesc' contient des exportations, on les garde.
if 'flowCode' in data.columns:
    # Si 'X' n'est pas le seul code pour export, on s'adapte.
    data = data[data['flowCode'].isin(['X', 'Exporter', 'Exports'])].copy()

# Suppression des agrégats pour éviter les doublons de niveaux
if 'isAggregate' in data.columns:
    data = data[data['isAggregate'] == False].copy()

print("Dimensions après filtrage :", data.shape)

# ------------------------------------------------------------
# 3. SELECTION ET RENOMMAGE DES VARIABLES
# ------------------------------------------------------------

data = data[
    ['refYear', 'reporterDesc', 'partnerDesc',
     'flowCode', 'cmdCode', 'cmdDesc',
     'primaryValue', 'netWgt']
].copy()

data = data.rename(columns={
    'refYear': 'Annee',
    'reporterDesc': 'Pays_declarant',
    'partnerDesc': 'Pays_partenaire',
    'flowCode': 'Flux',
    'cmdCode': 'Code_produit',
    'cmdDesc': 'Produit',
    'primaryValue': 'Valeur_USD',
    'netWgt': 'Poids_net_kg'
})

# ------------------------------------------------------------
# 4. CONVERSION DES TYPES
# ------------------------------------------------------------

data['Annee'] = pd.to_numeric(data['Annee'], errors='coerce')
data['Valeur_USD'] = pd.to_numeric(data['Valeur_USD'], errors='coerce')
data['Poids_net_kg'] = pd.to_numeric(data['Poids_net_kg'], errors='coerce')

data['Code_produit'] = data['Code_produit'].astype(str)
data['Produit'] = data['Produit'].astype(str)

# ------------------------------------------------------------
# 5. TRAITEMENT DES VALEURS MANQUANTES
# ------------------------------------------------------------

print("\nValeurs manquantes avant traitement :")
print(data.isnull().sum())

# Suppression des lignes sans année, produit ou valeur d'exportation
data = data.dropna(
    subset=['Annee', 'Code_produit', 'Produit', 'Valeur_USD']
).copy()

# ------------------------------------------------------------
# 6. SUPPRESSION DES DOUBLONS EXACTS
# ------------------------------------------------------------

nb_avant = len(data)

data = data.drop_duplicates()

print("\nNombre de doublons exacts supprimés :", nb_avant - len(data))

# ------------------------------------------------------------
# 7. VERIFICATION ET RESOLUTION DES DOUBLONS PRODUIT-ANNEE
# ------------------------------------------------------------

# S'il reste des doublons produit-année (par exemple différentes douanes ou partenaires),
# on agrège par produit et année pour garantir l'unicité avant de calculer t+1.
data = data.groupby(['Annee', 'Code_produit', 'Produit'], as_index=False).agg({
    'Valeur_USD': 'sum',
    'Poids_net_kg': 'sum'
})

print("\nDimensions après agrégation unique par Produit-Année :", data.shape)

# ------------------------------------------------------------
# 8. TRI CHRONOLOGIQUE
# ------------------------------------------------------------

data = data.sort_values(
    ['Code_produit', 'Annee']
).reset_index(drop=True)

# ------------------------------------------------------------
# 9. CREATION DE LA VARIABLE CIBLE (t+1)
# ------------------------------------------------------------

# Valeur des exportations du produit à l'année suivante
data['Annee_cible'] = (
    data.groupby('Code_produit')['Annee'].shift(-1)
)

data['Cible_t1'] = (
    data.groupby('Code_produit')['Valeur_USD'].shift(-1)
)

# Conservation des années consécutives uniquement
data['Ecart_annees'] = data['Annee_cible'] - data['Annee']

data_model = data[
    data['Ecart_annees'] == 1
].copy()

print("\nDimensions du jeu de modélisation :", data_model.shape)

# ------------------------------------------------------------
# 10. CONSTRUCTION DES VARIABLES X ET DE LA CIBLE y
# ------------------------------------------------------------

X = data_model[
    ['Annee_cible', 'Produit', 'Valeur_USD']
].copy()

X = X.rename(columns={
    'Valeur_USD': 'Valeur_annee_precedente_USD'
})

y = data_model['Cible_t1'].copy()

print("\nDimensions de X :", X.shape)
print("Dimensions de y :", y.shape)

# ------------------------------------------------------------
# 11. DECOUPAGE CHRONOLOGIQUE
# ------------------------------------------------------------

annees = sorted(data_model['Annee_cible'].unique())

if len(annees) < 5:
    raise ValueError(
        f"Nombre d'années insuffisant ({len(annees)} trouvées) pour créer "
        f"les ensembles train, validation et test. Veuillez vérifier vos filtres."
    )

# Les deux dernières années sont réservées au test final
annees_test = annees[-2:]

# Les deux années précédentes servent à la validation
annees_validation = annees[-4:-2]

# Les années les plus anciennes servent à l'entraînement
annees_train = annees[:-4]

train = data_model[
    data_model['Annee_cible'].isin(annees_train)
].copy()

validation = data_model[
    data_model['Annee_cible'].isin(annees_validation)
].copy()

test = data_model[
    data_model['Annee_cible'].isin(annees_test)
].copy()

X_train = X.loc[train.index]
X_val = X.loc[validation.index]
X_test = X.loc[test.index]

y_train = y.loc[train.index]
y_val = y.loc[validation.index]
y_test = y.loc[test.index]

print("\n===== DECOUPAGE DES DONNEES =====")

print("Années train :", annees_train)
print("Années validation :", annees_validation)
print("Années test :", annees_test)

print("\nTaille train :", X_train.shape)
print("Taille validation :", X_val.shape)
print("Taille test :", X_test.shape)

# ------------------------------------------------------------
# 12. PREPARATION DES VARIABLES SANS FUITE DE DONNEES
# ------------------------------------------------------------

variables_numeriques = [
    'Annee_cible',
    'Valeur_annee_precedente_USD'
]

variables_categorielles = ['Produit']

preparation = ColumnTransformer(
    transformers=[
        ('numerique', Pipeline([
            ('imputation', SimpleImputer(strategy='median')),
            ('standardisation', StandardScaler())
        ]), variables_numeriques),

        ('categorielle', Pipeline([
            ('imputation', SimpleImputer(strategy='most_frequent')),
            ('encodage', OneHotEncoder(handle_unknown='ignore'))
        ]), variables_categorielles)
    ]
)

# Apprentissage des transformations uniquement sur train
X_train_prepared = preparation.fit_transform(X_train)

# Application des transformations sur validation et test
X_val_prepared = preparation.transform(X_val)
X_test_prepared = preparation.transform(X_test)

# ------------------------------------------------------------
# 13. VERIFICATION FINALE
# ------------------------------------------------------------

print("\n===== VERIFICATION FINALE =====")

print("Train préparé :", X_train_prepared.shape)
print("Validation préparée :", X_val_prepared.shape)
print("Test préparé :", X_test_prepared.shape)

print("\nPréparation et découpage terminés avec succès.")
print("Les transformations ont été apprises uniquement sur train.")
```

## 5. Algorithme principal (un des 29 du cours)
**Algorithme / numéro :** Régression Ridge (Ridge Regression) — algorithme de régression linéaire régularisée.

**Pourquoi lui ? Alternative plausible écartée ?** La régression Ridge a été choisie parce que notre projet cherche à prédire une variable quantitative : la valeur des exportations agricoles marocaines en USD à l'année suivante (t+1).

Cet algorithme est adapté aux problèmes de régression et permet d'étudier la relation entre les exportations passées, l'année de prédiction et le type de produit.

Son principal avantage est la régularisation L2, qui réduit l'importance des coefficients trop élevés et contribue à limiter le surapprentissage.

Ce choix est particulièrement pertinent dans notre cas, car notre base de données contient un nombre limité d'observations exploitables.

Une alternative plausible serait le Random Forest Regressor, qui permet de capturer des relations non linéaires entre les variables.

Cependant, cet algorithme nécessite généralement davantage de données pour produire des résultats fiables. Avec seulement 12 observations d'entraînement, il risque de surapprendre.

La régression Ridge constitue donc un choix plus simple, interprétable et adapté à la taille actuelle de notre échantillon.

```python

# ============================================================
# PARTIE 5 : ALGORITHME PRINCIPAL - REGRESSION RIDGE
# ============================================================

from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd

# 1. Définition des hyperparamètres à tester
valeurs_alpha = [0.1, 1.0, 10.0]

# 2. Entraînement et comparaison sur validation
resultats_ridge = []

for alpha in valeurs_alpha:

    # Création du modèle
    modele_ridge = Ridge(
        alpha=alpha,
        fit_intercept=True
    )

    # Entraînement sur les données Train
    modele_ridge.fit(X_train_prepared, y_train)

    # Prédictions sur Validation
    predictions_val = modele_ridge.predict(X_val_prepared)

    # Calcul des métriques
    mae = mean_absolute_error(y_val, predictions_val)
    rmse = np.sqrt(mean_squared_error(y_val, predictions_val))

    wape = (
        np.sum(np.abs(y_val - predictions_val))
        / np.sum(np.abs(y_val))
    ) * 100

    resultats_ridge.append({
        'Alpha': alpha,
        'MAE': mae,
        'RMSE': rmse,
        'WAPE (%)': wape
    })

# 3. Affichage des résultats
tableau_ridge = pd.DataFrame(resultats_ridge)

print("===== COMPARAISON DES HYPERPARAMETRES RIDGE =====")
display(tableau_ridge)

# 4. Sélection du meilleur alpha selon la MAE
meilleur_alpha = tableau_ridge.loc[
    tableau_ridge['MAE'].idxmin(), 'Alpha'
]

print("Meilleur alpha :", meilleur_alpha)

# 5. Entraînement du modèle final de validation
modele_ridge = Ridge(
    alpha=meilleur_alpha,
    fit_intercept=True
)

modele_ridge.fit(X_train_prepared, y_train)

# 6. Prédictions sur Validation
predictions_ridge_val = modele_ridge.predict(X_val_prepared)

print("\nModèle Ridge entraîné avec succès.")
```
